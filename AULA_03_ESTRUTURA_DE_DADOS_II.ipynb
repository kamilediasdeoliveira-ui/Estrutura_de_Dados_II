```python
# PROBLEMA REAL - SISTEMA DE ATENDIMENTO DE UMA CLÍNICA
"""
Uma clínica recebe pacientes durante o dia. Cada paciente possui:

nome;
idade;
prioridade.

===================================

O sistema deve permitir:

adicionar paciente;
listar pacientes;
atender o primeiro paciente;
verificar se a fila está vazia;
informar a quantidade de pacientes.

"""
from dataclasses import dataclass

@dataclass
class Paciente:
  nome: str
  idade: int
  prioridade: str

class Node:
  def __init__(self,paciente):
    self.paciente = paciente
    self.proximo = None


class FilaAtendimento:
  def __init__(self):
    self.inicio = None
    self.fim = None


  def esta_vazia(self):
    if self.inicio is None:
      print("A fila está vazia")
      return True
    else:
      print("Há pacientes na fila")
      return False




  def adicionar(self):
    paciente = Paciente(
        nome = input (f"Nome: \n"),
        idade = int(input(f"Idade: \n")),
        prioridade = input(f"Prioridade: \n")
    )

    node = Node(paciente) # criando objeto node que recebe paciente


    if self.inicio is None:
      self.inicio = node
      self.fim = node
    else:
      self.fim.proximo = node
      self.fim = node


  def listar(self):
    atual = self.inicio

    if atual is None:
      print("A fila está vazia\n")

    while atual is not None:

      p = atual.paciente  # acessa o ogjeto  paciente dentro do nó
      print(f"Nome:{p.nome} | Idade: {p.idade} | prioridade {p.prioridade}")


      atual = atual.proximo


  def atender(self):
    # atender primeiro paciente

    atual = self.inicio

    if atual is None:
        print("A fila está vazia\n")
        return

    p = atual.paciente
    print(f"Paciente {p.nome} atendido")

    if self.inicio == self.fim:
        self.inicio = None
        self.fim = None
    else:
        self.inicio = atual.proximo


  def tamanho(self):

    quantidade = 0

    atual = self.inicio

    if atual is None:
      print("A fila está vazia\n")

    while atual is not None:
      quantidade += 1
      atual = atual.proximo

    print(f"Quandidade de pessoas na fila : {quantidade}")


def menu():

  opcao = -1

  while True:
    print ("\n ==== MENU ===\n")
    print("1 - Adicionar paciente")
    print("2 - Listar pacientes")
    print("3 - Atender primeiro paciente")
    print("4 - Verificar se a fila está vazia")
    print("5 - Informar quantidade de pacientes")
    print("6 - Sair")

    opcao = int(input("Escolha uma opção: "))

    if opcao == 1:
      fila.adicionar()
    elif opcao == 2:
      fila.listar()
    elif opcao == 3:
      fila.atender()
    elif opcao == 4:
      fila.esta_vazia()
    elif opcao == 5:
      fila.tamanho()
    elif opcao == 6:
      print("Programa encerrado")
      break
    else:
      print("Opção inváliada")




fila = FilaAtendimento()
menu()

```

```python
# PROBLEMA REAL - SISTEMA DE ATENDIMENTO DE UMA CLÍNICA - VERSÃO FINAL

"""
Amplie o sistema.
Pacientes com prioridade devem ser atendidos antes dos pacientes normais.

Teste:
fila vazia;
um paciente;
vários pacientes;
atendimento;
remoção do último paciente;
paciente prioritário.
"""
from dataclasses import dataclass

@dataclass
class Paciente:
  nome: str
  idade: int

  def __post_init__(self):
    self.prioridade = self.idade >= 60

class Node:
  def __init__(self,paciente):
    self.paciente = paciente
    self.proximo = None


class FilaAtendimento:
  def __init__(self):
    self.inicio = None
    self.fim = None
    self._tamanho = 0


  def esta_vazia(self):
    if self.inicio is None:
      print("A fila está vazia")
      return True
    else:
      print("Há pacientes na fila")
      return False





  def adicionar(self):
    paciente = Paciente(
        nome = input (f"Nome: "),
        idade = int(input(f"Idade: ")),
    )
    node = Node(paciente) # criando objeto node que recebe paciente
    p = node.paciente

    if self.inicio is None:
      self.inicio = node
      self.fim = node
    else:
      if p.prioridade:

        # O primeiro paciente é normal
        if not self.inicio.paciente.prioridade:
            node.proximo = self.inicio
            self.inicio = node

        else:
          # Já existem prioritários
            atual = self.inicio

        # Procura o último paciente prioritário
            while (atual.proximo is not None and
                  atual.proximo.paciente.prioridade):
              atual = atual.proximo

            # Insere o novo prioritário depois dos prioritários existentes
            node.proximo = atual.proximo
            atual.proximo = node

            # Se entrou no final, atualiza o fim
            if node.proximo is None:
              self.fim = node

      else:
        self.fim.proximo = node
        self.fim = node

    self._tamanho += 1

  def listar(self):
    atual = self.inicio

    if atual is None:
      print("A fila está vazia\n")
      return

    while atual is not None:

      p = atual.paciente  # acessa o ogjeto  paciente dentro do nó

      if p.prioridade:
        tipo = "Prioridade"
      else:
        tipo = "Normal"

      print(f"Nome:{p.nome} | Idade: {p.idade} | prioridade {tipo}")


      atual = atual.proximo


  def atender(self):
    # atender primeiro paciente

    atual = self.inicio

    if atual is None:
        print("A fila está vazia\n")
        return

    p = atual.paciente
    print(f"Paciente {p.nome} atendido")

    if self.inicio == self.fim:
        self.inicio = None
        self.fim = None
    else:
        self.inicio = atual.proximo

    self._tamanho -= 1


  def tamanho(self):

    print(f"Quantidade de pessoas na fila : {self._tamanho}")


def menu():

  while True:
    print ("\n ==== MENU ===\n")
    print("1 - Adicionar paciente")
    print("2 - Listar pacientes")
    print("3 - Atender primeiro paciente")
    print("4 - Verificar se a fila está vazia")
    print("5 - Informar quantidade de pacientes")
    print("6 - Sair")

    opcao = int(input("Escolha uma opção: "))

    if opcao == 1:
      fila.adicionar()
    elif opcao == 2:
      fila.listar()
    elif opcao == 3:
      fila.atender()
    elif opcao == 4:
      fila.esta_vazia()
    elif opcao == 5:
      fila.tamanho()
    elif opcao == 6:
      print("Programa encerrado")
      break
    else:
      print("Opção inváliada")




fila = FilaAtendimento()
menu()



```

O que é uma referência em Python?

R = É uma variável que aponta para um objeto na memória. Assim, duas variáveis podem fazer referência ao mesmo objeto

Qual a diferença entre b = ae b = a.copy()?

R = b = a faz b apontar para o mesmo objeto que a. Já b = a.copy() cria uma cópia do objeto, permitindo que b seja alterado sem modificar o objeto original em alguns casos

Por que uma estrutura encadeada precisa de referências?

R = Porque as referências permitem que cada nó saiba onde está o próximo nó

Qual erro você encontrou durante a depuração?

R = Laço infiniito no listar()

Erro:

while atual is not None: p = atual.paciente

o método listar() entrava em um laço infinito porque a variável atual não avançava para o próximo nó.

Correção: adicionei atual = atual.proximo dentro do while, permitindo que a fila seja percorrida até chegar em None.

while atual is not None: p = atual.paciente print(...) atual = atual.proximo

Como referências ajudam a compreender estruturas dinâmicas?

R = Elas ajudam a entender como os nós são ligados entre si. Na fila, por exemplo, proximo guarda a referência para o próximo nó
