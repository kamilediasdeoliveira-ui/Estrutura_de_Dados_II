```python
# EXERCÍCIO 2 - CONSTRUINDO UMA CADEIA

class Node:
  def __init__(self,valor):
    self.valor = valor
    self.proximo = None


n1 = Node("A")
n2 = Node("B")
n3 = Node("C")

n1.proximo = n2
n2.proximo = n3

atual = n1
while atual is not None:
    print(atual.valor)
    atual = atual.proximo


```

```python
# EXERCÍCIO 3 - DEPURAÇÃO

n1 = Node(10)
n2 = Node(20)
n3 = Node(30)

n1.proximo = n2
n2.proximo = n3

atual = n1

while atual is not None:
    print(atual.valor)
```
