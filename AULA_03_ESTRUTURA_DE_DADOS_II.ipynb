```python
# EXERCÍCIO 2 - CONSTRUINDO UMA CADEIA

class Node:
  def __init__(self,valor):
    self.valor = valor
    self.proximo = None


n1 = Node("A")
n2 = Node("B")
n3 = Node("C")

n1.proximo = n2
n2.proximo = n3

atual = n1
while atual is not None:
    print(atual.valor)
    atual = atual.proximo


```

```python
# PROBLEMA REAL - SISTEMA DE ATENDIMENTO DE UMA CLÍNICA
"""
Uma clínica recebe pacientes durante o dia. Cada paciente possui:

nome;
idade;
prioridade.

===================================

O sistema deve permitir:

adicionar paciente;
listar pacientes;
atender o primeiro paciente;
verificar se a fila está vazia;
informar a quantidade de pacientes.

"""
from dataclasses import dataclass

@dataclass
class Paciente:
  nome: str
  idade: int
  prioridade: str

class Node:
  def __init__(self,paciente):
    self.paciente = paciente
    self.proximo = None


class FilaAtendimento:
  def __init__(self):
    self.inicio = None
    self.fim = None


  def esta_vazia(self):
    if self.inicio is None:
      print("A fila está vazia")
      return True
    else:
      print("Há pacientes na fila")
      return False




  def adicionar(self):
    paciente = Paciente(
        nome = input (f"Nome: \n"),
        idade = int(input(f"Idade: \n")),
        prioridade = input(f"Prioridade: \n")
    )

    node = Node(paciente) # criando objeto node que recebe paciente


    if self.inicio is None:
      self.inicio = node
      self.fim = node
    else:
      self.fim.proximo = node
      self.fim = node


  def listar(self):
    atual = self.inicio

    if atual is None:
      print("A fila está vazia\n")

    while atual is not None:

      p = atual.paciente  # acessa o ogjeto  paciente dentro do nó
      print(f"Nome:{p.nome} | Idade: {p.idade} | prioridade {p.prioridade}")


      atual = atual.proximo


  def atender(self):
    # atender primeiro paciente

    atual = self.inicio

    if atual is None:
        print("A fila está vazia\n")
        return

    p = atual.paciente
    print(f"Paciente {p.nome} atendido")

    if self.inicio == self.fim:
        self.inicio = None
        self.fim = None
    else:
        self.inicio = atual.proximo


  def tamanho(self):

    quantidade = 0

    atual = self.inicio

    if atual is None:
      print("A fila está vazia\n")

    while atual is not None:
      quantidade += 1
      atual = atual.proximo

    print(f"Quandidade de pessoas na fila : {quantidade}")


def menu():

  opcao = -1

  while True:
    print ("\n ==== MENU ===\n")
    print("1 - Adicionar paciente")
    print("2 - Listar pacientes")
    print("3 - Atender primeiro paciente")
    print("4 - Verificar se a fila está vazia")
    print("5 - Informar quantidade de pacientes")
    print("6 - Sair")

    opcao = int(input("Escolha uma opção: "))

    if opcao == 1:
      fila.adicionar()
    elif opcao == 2:
      fila.listar()
    elif opcao == 3:
      fila.atender()
    elif opcao == 4:
      fila.esta_vazia()
    elif opcao == 5:
      fila.tamanho()
    elif opcao == 6:
      print("Programa encerrado")
      break
    else:
      print("Opção inváliada")




fila = FilaAtendimento()
menu()


#fila.adicionar(Paciente("Bruno", 70, "Prioridade"))
#fila.adicionar(Paciente("Carlos", 45, "Normal"))

#fila.listar()

#paciente = fila.atender()
#print("Atendendo:", paciente.nome)



```
