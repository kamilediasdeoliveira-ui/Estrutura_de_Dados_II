```python
# EXERCÍCIO 2 - CONSTRUINDO UMA CADEIA

class Node:
  def __init__(self,valor):
    self.valor = valor
    self.proximo = None


n1 = Node("A")
n2 = Node("B")
n3 = Node("C")

n1.proximo = n2
n2.proximo = n3

atual = n1
while atual is not None:
    print(atual.valor)
    atual = atual.proximo


```

```python
# EXERCÍCIO 3 - DEPURAÇÃO

n1 = Node(10)
n2 = Node(20)
n3 = Node(30)

n1.proximo = n2
n2.proximo = n3

atual = n1

while atual is not None:
    print(atual.valor)
```

```python
# PROBLEMA REAL - SISTEMA DE ATENDIMENTO DE UMA CLÍNICA
"""
Uma clínica recebe pacientes durante o dia. Cada paciente possui:

nome;
idade;
prioridade.

===================================

O sistema deve permitir:

adicionar paciente;
listar pacientes;
atender o primeiro paciente;
verificar se a fila está vazia;
informar a quantidade de pacientes.

"""
from dataclasses import dataclass

@dataclass
class Paciente:
  nome: str
  idade: int
  prioridade: str

class Node:
  def __init__(self,paciente):
    self.paciente = paciente
    self.proximo = None


class FilaAtendimento:
  def __init__(self):
    self.inicio = None
    self.fim = None


  def adicionar(self):
    paciente = Paciente(
        nome = input (f"Nome: \n"),
        idade = int(input(f"Idade: \n")),
        prioridade = input(f"Prioridade: \n")
    )

    node = Node(paciente) # criando objeto node que recebe paciente


    if self.inicio is None:
      self.inicio = node
      self.fim = node
    else:
      self.fim.proximo = node
      self.fim = node

fila = FilaAtendimento()



```
